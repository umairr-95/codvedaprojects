```python
# Level 1 Task 3: Basic Data Visualization

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("D:/Codveda Technologies project/Umair-L1-TASK 3 Basic Data visualization/1) iris.csv")
```

```python
df.head()
```

```python
print("Dataset shape:", df.shape)

print("/nColumn names:")
print(df.columns)

print("/nData types:")
print(df.dtypes)
```

```python
# Bar Chart

species_count = df['species'].value_counts()

plt.figure(figsize=(8, 5))
species_count.plot(kind='bar')

plt.title('Number of Iris by Species')
plt.xlabel('Species')
plt.ylabel('Number of petals')

plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig('species_bar_plot.png')
plt.show()
```

<!-- The number of observations for each species of iris is displayed in the bar plot. It makes it easy to compare how many records each species in the dataset has. -->

```python
# scatter plot

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='petal_length',
    y='petal_width',
    hue='species'
)

plt.title('Petal Length vs Petal Width')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')

plt.legend(title='Species')
plt.tight_layout()

plt.savefig('petal_length_vs_width.png')
plt.show()
```

<!-- The variance in sepal length across the observations is displayed in the line chart. It is possible to visualise the fluctuation in sepal length because the values vary from one observation to the next. -->

```python
# Line Chart

plt.figure(figsize=(10, 5))

plt.plot(df.index, df['sepal_length'], label='Sepal Length')

plt.title('Sepal Length Observations')
plt.xlabel('Observation')
plt.ylabel('Sepal Length')

plt.legend()
plt.tight_layout()

plt.savefig('sepal_length_line_chart.png')
plt.show()
```

<!-- The correlation between petal width and length is displayed in the scatter plot. It is feasible to visually compare the measurements between the various species because the points are represented by species. -->
